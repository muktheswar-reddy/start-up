```python
# 4. Import and verify
import torch
import transformers
import datasets
import numpy as np
import pandas as pd
from scipy import stats
import sklearn
import matplotlib
import pytest

print('transformers:', transformers.__version__)
print('datasets:', datasets.__version__)
print('numpy:', np.__version__)
print('pandas:', pd.__version__)
print('sklearn:', sklearn.__version__)
print('matplotlib:', matplotlib.__version__)
```

```python
# 5. Set seed
SEED = 42
import random
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
print(f'Seed set to {SEED}')
```

```python
# 6. Create output directories
Path('results').mkdir(exist_ok=True)
Path('reports').mkdir(exist_ok=True)
Path('data').mkdir(exist_ok=True)

# 7. Environment report
env_report = {
    'python_version': sys.version,
    'platform': platform.platform(),
    'torch_version': torch.__version__,
    'cuda_available': torch.cuda.is_available() if torch.cuda.is_available() else None,
    'transformers_version': transformers.__version__,
    'numpy_version': np.__version__,
    'seed': SEED,
}

Path('results/environment.json').write_text(json.dumps(env_report, indent=2))
print('Environment report saved to results/environment.json')
env_report
```
